```python
import pandas as pd
import os
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
dataEncod = pd.read_csv('../data/processed/dataset_clean.csv')
dataEncod = dataEncod.drop(columns=['gpa_category'])
```

```python


plt.figure(figsize=(14, 10))
matriz_corr = dataEncod.corr(method='pearson')

sns.heatmap(
    matriz_corr,
    annot=True,
    fmt='.2f',
    cmap='coolwarm',
    vmin=-1,
    vmax=1,
    linewidths=0.5
)
plt.title('Matriz de Correlação Pearson (Todas as Variáveis Codificadas)', fontsize=14)
plt.tight_layout()
os.makedirs('../reports/figures', exist_ok=True)
plt.savefig('../reports/figures/matriz_correlacao_regressao.png', bbox_inches='tight', dpi=300)
plt.show()
```

```python
col_util = ['avg_sleep_hours', 'sleep_midpoint_minutes', 'prior_gpa', 'underrepresented', 'first_generation', 'term_gpa']

dataRegression = dataEncod[col_util].copy()
```

```python
caminho_dados_processados = '../data/processed/'
os.makedirs(caminho_dados_processados, exist_ok=True)
dataRegression.to_csv(f'{caminho_dados_processados}dataset_regression.csv', index=False)
```
