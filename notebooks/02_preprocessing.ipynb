```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
import os
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler 
from scipy.stats.mstats import winsorize
from pathlib import Path
import sys

pasta_principal = str(Path().resolve().parent)
if pasta_principal not in sys.path:
    sys.path.append(pasta_principal)

from src.data_loader import carregar_dados_brutos
from src.preprocessing import (
    corrigir_valores_invalidos,
    filtrar_colunas,
    transformar_categoricas_em_numeros,
)
```

```python
dataLimpo = carregar_dados_brutos('../data/raw/college_sleep_and_gpa.csv')
```

```python
invalidos = dataLimpo[~dataLimpo['first_generation'].isin([0, 1]) & dataLimpo['first_generation'].notna()]['first_generation']

if not invalidos.empty:
    for val in invalidos.unique():
        print(f"Valor inválido encontrado na coluna 'first_generation': {val}")
    
    dataLimpo = corrigir_valores_invalidos(dataLimpo)
    print("O valor 2 foi substituído por 1 com sucesso.")
```

```python
colInutil = [
    'avg_sleep_minutes',
    'term_units',
    'student_id',
    'sleep_midpoint_clock',
    'cohort_code',
    'study',
    'semester',
]
colunas_mantidas = [coluna for coluna in dataLimpo.columns if coluna not in colInutil]
dataLimpo = filtrar_colunas(dataLimpo, colunas_mantidas)
```

```python
dataLimpo = dataLimpo.dropna(subset=['gender', 'first_generation', 'underrepresented'])
```

```python
dataLimpo['term_load_z'] = dataLimpo['term_load_z'].fillna(dataLimpo['term_load_z'].median())
print(dataLimpo.isnull().sum())
```

```python
transfLog =['bedtime_variability', 'daytime_sleep_minutes']

for col in transfLog:
    dataLimpo[col] = np.log1p(dataLimpo[col])

dataLimpo['sleep_midpoint_minutes'] = winsorize(dataLimpo['sleep_midpoint_minutes'], limits=[0, 0.03])

dataLimpo['term_load_z'] = winsorize(dataLimpo['term_load_z'], limits=[0.03, 0.03])
```

```python
mediana_gpa = dataLimpo['term_gpa'].median()
dataLimpo['gpa_category'] = np.where(dataLimpo['term_gpa'] >= mediana_gpa, 'Alto', 'Baixo')

print(dataLimpo['gpa_category'].value_counts())
```

```python
dataEncod = dataLimpo.copy()
```

```python
print(dataLimpo.isnull().sum())
```

```python
bracketsEncod = {
    '<5.5h': 1,
    '5.5-6h': 2,
    '6-6.5h': 3,
    '6.5-7h': 4,
    '7-7.5h': 5,
    '7.5h+': 6,
}

dataEncod['sleep_bracket_encoded'] = dataEncod['sleep_bracket'].map(bracketsEncod)
dataEncod = dataEncod.drop(columns=['sleep_bracket'])
```

```python
dataEncod['gpa_category_encoded'] = dataEncod['gpa_category'].map({'Alto': 1, 'Baixo': 0})
```

```python
print(dataEncod.isnull().sum())
```

```python

dataEncod = transformar_categoricas_em_numeros(
    dataEncod, colunas=['gender', 'university']
)

dataEncod.head()
```

```python
dataLimpo = dataEncod.copy()
```

```python
scaler = StandardScaler()

colunas_continuas = [
    'avg_sleep_hours',
    'daytime_sleep_minutes',
    'sleep_midpoint_minutes',
    'bedtime_variability',
    'nights_tracked_fraction',
    'prior_gpa',
    'gpa_change',
    'term_load_z'
]

dataEncod[colunas_continuas] = scaler.fit_transform(dataEncod[colunas_continuas])

dataEncod.describe()
```

```python
caminho_dados_processados = '../data/processed/'
os.makedirs(caminho_dados_processados, exist_ok=True)
dataEncod.to_csv(f'{caminho_dados_processados}dataset_clean_encoded.csv', index=False)
```

```python
dataEncod = pd.read_csv('../data/processed/dataset_clean_encoded.csv')
```

```python
grupo_alto = dataEncod[dataEncod['gpa_category'] == 'Alto']
grupo_baixo = dataEncod[dataEncod['gpa_category'] == 'Baixo']
features = ['avg_sleep_hours', 'sleep_midpoint_minutes', 'prior_gpa', 'underrepresented', 'first_generation']

print("=== TESTE T DE STUDENT (Alto vs Baixo Desempenho) ===")
for col in features:
    stat, p_val = stats.ttest_ind(grupo_alto[col], grupo_baixo[col])
    significativo = "Significativo (p < 0.05)" if p_val < 0.05 else "Não significativo"
    print(f"{col:25s} | p-valor: {p_val:.4f} -> {significativo}")
```

```python
SEED = 42

dataReg = pd.read_csv('../data/processed/dataset_regression.csv')
dataClass = pd.read_csv('../data/processed/dataset_classification.csv')

X_reg = dataReg.drop(columns=['term_gpa'])
y_reg = dataReg['term_gpa']

X_class = dataClass.drop(columns=['gpa_category_encoded'])
y_class = dataClass['gpa_category_encoded']
```

```python
X_train_reg, X_test_reg, y_train_reg, y_test_reg = train_test_split(
    X_reg, y_reg, test_size=0.20, random_state=SEED
)

# Classificação (Com Estratificação)
X_train_class, X_test_class, y_train_class, y_test_class = train_test_split(
    X_class, y_class, test_size=0.20, random_state=SEED, stratify=y_class
)

print(f"Regressão - Treino: {X_train_reg.shape[0]} amostras | Teste: {X_test_reg.shape[0]} amostras")
print(f"Classificação - Treino: {X_train_class.shape[0]} amostras | Teste: {X_test_class.shape[0]} amostras")
```

```python
col_util = ['avg_sleep_hours', 'sleep_midpoint_minutes', 'prior_gpa', 'underrepresented', 'first_generation']
dataLimpoClass = dataLimpo[col_util].copy()

# O .values ignora a checagem de index e cola na ordem das linhas:
dataLimpoClass['gpa_category_encoded'] = dataEncod['gpa_category_encoded'].values

caminho_dados_processados = '../data/processed/'
os.makedirs(caminho_dados_processados, exist_ok=True)
dataLimpoClass.to_csv(f'{caminho_dados_processados}data_limpo_class.csv', index=False)
```

```python
ColNumReg = X_reg.select_dtypes(include=[np.number]).columns
ColNumClass = X_class.select_dtypes(include=[np.number]).columns

# Criando os transformadores de colunas
pipelineReg = ColumnTransformer(
    transformers=[
        ('scaler', StandardScaler(), ColNumReg)
    ],
    remainder='passthrough'
)

pipelineClass = ColumnTransformer(
    transformers=[
        ('scaler', StandardScaler(), ColNumClass)
    ],
    remainder='passthrough'
)

X_train_reg_scaled = pipelineReg.fit_transform(X_train_reg)
X_test_reg_scaled = pipelineReg.transform(X_test_reg)

X_train_class_scaled = pipelineClass.fit_transform(X_train_class)
X_test_class_scaled = pipelineClass.transform(X_test_class)

print("✅ Protocolo experimental e Pipelines configurados com sucesso!")
```
